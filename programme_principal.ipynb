```python
import json
from typing import Dict, List, Any

# On récupère les 4 fonctions du fichier fonctions_salaire_entreprise.py.
from fonctions_salaire_entreprise import (
    calculer_salaire_employe,
    afficher_statistiques_globales,
    afficher_stats_une_filiale,
    creer_csv
)


# On ouvre notre fichier JSON.
with open("employes_data_test.json", "r", encoding="utf-8") as fichier:
    donnees_entreprise: Dict[str, Any] = json.load(fichier)


# Crée une liste pour les salaires, 
# Créer un dict pour les filiales.
liste_tous_les_salaires: List[float] = []
dictionnaire_par_filiale: Dict[str, List[float]] = {}


# On parcourt les filiales une par une.
for nom_filiale, liste_employes in donnees_entreprise.items():

    # On crée une liste vide pour chaque filiale.
    dictionnaire_par_filiale[nom_filiale] = []

    # On affiche le nom de la filiale.
    print("\n" + "=" * 85)
    print(f" STRUCTURE : {nom_filiale.upper()}")
    print("=" * 85)
    print("  NOM        |  POSTE          |  SALAIRE MENSUEL")
    print("-" * 85)

    # On parcourt les employés de la filiale.
    for employe in liste_employes:

        # On calcule le salaire de l'employé.
        salaire_mensuel: float = calculer_salaire_employe(employe)

        # On garde le salaire pour les stats.
        liste_tous_les_salaires.append(salaire_mensuel)
        dictionnaire_par_filiale[nom_filiale].append(salaire_mensuel)

        # On affiche l'employé, son poste et son salaire.
        print(
            f"  {employe['name']:<10} | "
            f"  {employe['job']:<14} | "
            f"  {round(salaire_mensuel, 2)} €"
        )

    print("-" * 85)


print("\n\n")


# On affiche les statistiques de toute l'entreprise.
afficher_statistiques_globales(liste_tous_les_salaires)


# On affiche les statistiques de chaque filiale.
print("-" * 65)
print("             STATISTIQUES PAR FILIALE")
print("-" * 65)

# On parcourt les filiales avec leurs salaires.
for nom_filiale, liste_salaires in dictionnaire_par_filiale.items():

    # On affiche les statistiques de la filiale.
    afficher_stats_une_filiale(nom_filiale, liste_salaires)


# Appel de la fonction D.
creer_csv(donnees_entreprise)
```
